# EX: Formulating an MDP

In this exercise, we will formulate the stochastic cyber-defense scenario using Python dictionaries.

The Scenario: A network agent detects a Suspicious_Connection. It must decide whether to Block the connection or actively Monitor it to gather intelligence.

We will program the Transition Function $P(s' \mid s,a)$, assign the Reward Function $R(s,a,s')$, and calculate the mathematical Expected Value (EV) of choosing to Monitor versus Block.

## Steps to formulate this in Python:

1. Create a nested dictionary mapping the Current State -> Chosen Action -> List of Possible Outcomes.

2. Each possible outcome must be a tuple containing (Probability, Next_State, Reward).

3. Create an evaluation function that loops through these outcomes, multiplies the probability by the reward, and sums them up to find the Expected Immediate Reward.




```python

# Step 1: Define the Transition Function P(s' | s, a) and Reward Function
# Format: transitions[state][action] = [(Probability, Next_State, Reward)]
transitions = {
    "Suspicious_Connection": {
        "Block": [
            (0.95, "Secure", +10),        # 95% chance the block succeeds
            (0.05, "Compromised", -100)   # 5% chance the adversary bypasses the firewall
        ],
        "Monitor": [
            (0.60, "Secure", +5),         # 60% chance the connection is benign
            (0.40, "Compromised", -100)   # 40% chance the monitored connection drops a payload
        ]
    }
}

# Step 2: Define a function to calculate the Expected Value of an action
def calculate_expected_immediate_reward(state, action):
    expected_value = 0.0
    outcomes = transitions[state][action]
    
    print(f"--- Tactical Evaluation: {action} a {state} ---")
    
    # Step 3: Loop through outcomes and calculate Expected Value
    for prob, next_state, reward in outcomes:
        ev_branch = prob * reward
        expected_value += ev_branch
        print(f"Outcome: {next_state} | Prob: {prob*100}% | Reward: {reward} | EV Contribution: {ev_branch}")
        
    print(f">> Total Expected Immediate Reward: {expected_value}\n")
    return expected_value

# Execute the lab
calculate_expected_immediate_reward("Suspicious_Connection", "Block")
calculate_expected_immediate_reward("Suspicious_Connection", "Monitor")
```

## Interpreting the Results

When executing this code, the system calculates the mathematical Expected Value (EV) of each action by multiplying the probabilities by their respective rewards.

For Block, the math is (0.95 * 10) + (0.05 * -100) = 9.5 - 5.0 = +4.5.

For Monitor, the math is (0.6 * 5) + (0.4 * -100) = 3.0 - 40.0 = -37.0.

Even though monitoring traffic might yield valuable intelligence, the MDP mathematically proves that performing the Block action is the optimal policy from this state, as it drastically minimizes the expected risk of compromise.
