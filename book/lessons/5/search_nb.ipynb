# EX: Frontier Data Structures

The math driving these distinct tactical behaviors comes entirely from the data structure used for the Frontier. In this lab, we will simulate how Python handles the Stack (DFS), Queue (BFS), and Priority Queue (UCS).


````python
from collections import deque
import heapq

print("--- 1. DFS: The LIFO Stack ---")
# In Python, a standard list acts as a Stack using append() and pop()
dfs_frontier = []
dfs_frontier.append("Path_A (Depth 1)")
dfs_frontier.append("Path_B (Depth 2)")
dfs_frontier.append("Path_C (Depth 3)")

# pop() removes the LAST item added (Deepest)
next_node = dfs_frontier.pop() 
print(f"DFS expands: {next_node}\n")


print("--- 2. BFS: The FIFO Queue ---")
# In Python, we use deque for an efficient FIFO Queue
bfs_frontier = deque()
bfs_frontier.append("Path_A (Tier 1)")
bfs_frontier.append("Path_B (Tier 1)")
bfs_frontier.append("Path_C (Tier 2)")

# popleft() removes the FIRST item added (Shallowest)
next_node = bfs_frontier.popleft() 
print(f"BFS expands: {next_node}\n")


print("--- 3. UCS: The Priority Queue ---")
# In Python, we use heapq. We store tuples of (Cumulative_Cost, Node_Name)
ucs_frontier = []
# heapq automatically sorts based on the first item in the tuple (the cost)
heapq.heappush(ucs_frontier, (10.0, "Path_A"))
heapq.heappush(ucs_frontier, (2.5,  "Path_B")) # Cheapest!
heapq.heappush(ucs_frontier, (15.0, "Path_C"))

# heappop() always removes the item with the LOWEST cost
cost, next_node = heapq.heappop(ucs_frontier)
print(f"UCS expands: {next_node} with cost {cost}\n")
````
