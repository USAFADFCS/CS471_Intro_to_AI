# EX: Evaluating MDP Action Returns

In this exercise, we will use Python to automate the decision-making process of an AI system comparing two different tactical actions based on their expected discounted returns.

An automated cyber-defense agent detects anomalous traffic. It must evaluate two potential countermeasures:
*   **Action A (Quarantine):** Yields an immediate reward of $+4$ and successfully terminates the threat (ends the episode).
*   **Action B (Observe & Log):** Yields a small immediate reward of $+1$ now, and a larger reward of $+4$ on the next step when the adversary's IP is fully traced. 
*   The discount factor for the system is $\gamma = 0.5$.

Which action should the AI mathematically prefer?


```python
def calculate_return(rewards, gamma):
    """
    Calculates the expected discounted return for a sequence of rewards.
    """
    total_return = 0.0
    for t, r in enumerate(rewards):
        total_return += (gamma ** t) * r
    return total_return

# Define the discount factor
gamma = 0.5

# Define the sequence of rewards for each action choice
# Action A: +4 immediately, then terminates (no further rewards)
rewards_A = [4]

# Action B: +1 immediately, +4 on the next step
rewards_B = [1, 4]

# Calculate the expected returns
return_A = calculate_return(rewards_A, gamma)
return_B = calculate_return(rewards_B, gamma)

print(f"Expected Return for Action A (Quarantine): {return_A}")
print(f"Expected Return for Action B (Observe & Log): {return_B}")

# Agent Decision Logic
if return_A > return_B:
    print("Optimal Choice: Execute Action A")
elif return_B > return_A:
    print("Optimal Choice: Execute Action B")
else:
    print("Actions are mathematically equal in value.")
```

## Interpreting the Results

Even though Action B technically offers a higher raw sum of points ($1 + 4 = 5$), the severe discount factor ($\gamma = 0.5$) drastically reduces the value of the delayed $+4$ reward down to $+2$ ($1 + 0.5(4) = 3$).

Therefore, the greedy optimal choice for the cyber agent is Action A.
