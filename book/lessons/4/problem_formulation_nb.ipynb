# EX: State, Goal Test, and Successor Function

In this lab, we won't solve the maze yet. Instead, we will build the mathematical formulation of a drone navigating a contested grid. We will define the State, the Goal Test, and the Successor Function.

**Steps Performed:**

* **Define the physical environment** (a 2D grid with a no-fly zone).

* **Create a class** to represent the Search Problem.

* **Implement logic** for the get_successors function to calculate the valid next states and their path costs.

* **Pass a test state** into the successor function to observe the AI's "Fringe" options.


````python
class TacticalRoutingProblem:
    """A formal representation of a grid-based search problem."""
    def __init__(self, grid_size, start, goal, no_fly_zones):
        self.max_x, self.max_y = grid_size
        self.start_state = start
        self.goal_state = goal
        self.no_fly_zones = set(no_fly_zones)

    def get_start_state(self):
        return self.start_state

    def is_goal_state(self, state):
        return state == self.goal_state

    def get_successors(self, state):
        """
        The Successor Function.
        Returns a list of tuples: (next_state, action, path_cost)
        """
        successors = []
        x, y = state
        
        # Standard Actions: North, South, East, West
        # Cost is 1.0 for each movement.
        actions = {
            "North": (x, y + 1, 1.0),
            "South": (x, y - 1, 1.0),
            "East":  (x + 1, y, 1.0),
            "West":  (x - 1, y, 1.0)
        }
        
        for action_name, (next_x, next_y, cost) in actions.items():
            # Abstraction Rule 1: Stay within the grid
            if 0 <= next_x <= self.max_x and 0 <= next_y <= self.max_y:
                # Abstraction Rule 2: Avoid no-fly zones
                if (next_x, next_y) not in self.no_fly_zones:
                    successors.append(((next_x, next_y), action_name, cost))
                    
        return successors

# --- Scenario ---
# 5x5 grid, Start at (0,0), Goal at (4,4), SAM threat at (1,0) and (1,1)
mission = TacticalRoutingProblem(
    grid_size=(4, 4), 
    start=(0, 0), 
    goal=(4, 4), 
    no_fly_zones=[(1, 0), (1, 1)]
)

print(f"Start State: {mission.get_start_state()}")
print(f"Goal State: {mission.goal_state}")

# Let's test the successor function from the Start State
options = mission.get_successors(mission.get_start_state())
print(f"\nOptions from Start {mission.get_start_state()}:")
for next_state, action, cost in options:
    print(f" - Action: {action} -> Results in State: {next_state} (Cost: {cost})")

````

When you execute this script, you are watching the AI evaluate its Fringe. From state (0,0), mathematically, it should have two options: North (0,1) and East (1,0).

However, because we formulated (1,0) as a no-fly zone (representing a threat ring), the get_successors function filters it out. The AI only outputs North (0,1). By perfectly abstracting the physical rules of the world into this mathematical function, we guarantee that any search algorithm (DFS, BFS, A*) that uses this function will never plot a path through the SAM threat.
