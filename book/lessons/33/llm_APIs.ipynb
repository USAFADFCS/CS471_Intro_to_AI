## EX: Interacting with LLM APIs

Instead of building a massive LLM from scratch (which costs millions of dollars), the Air Force accesses highly capable Instruct-Tuned models via an {term}`API (Application Programming Interface)`.

In this exercise, we will use Python to send an automated HTTP request to an LLM endpoint, asking it to summarize a lengthy, chaotic AAR.

* **Prepare the Payload:** Format the AAR text and our explicit instructions (the "System Prompt").

* **Call the API:** Send the data to the LLM securely using Python's `requests` library.

* **Parse the Output:** Extract the actionable intelligence summary from the JSON response.


```python
# Only run this cell after downloading and selecting your kernel
!python.exe -m pip install --upgrade pip
!pip install numpy matplotlib scikit-learn pandas tensorflow requests
```

```python

import requests
import json

# 1. Define the API Endpoint and Authentication
# Note: In a real environment, NEVER hardcode API keys. Use environment variables.
API_URL = "https://generativelanguage.googleapis.com/v1beta/models/gemini-3-flash-preview:generateContent"
API_KEY = "YOUR_API_KEY_HERE"  

# 2. The Tactical Data (After Action Report)
aar_text = """
At 0400 hours, Bravo Company initiated movement towards Grid Target Alpha. 
Heavy rain degraded visual ISR capabilities. At 0530, lead elements reported 
small arms contact from the ridgeline. Air support was requested but denied 
due to weather. Bravo Company successfully flanked the adversarial position 
using the western ravine. Two minor injuries reported. Target secured at 0715.
"""

# 3. Construct the Payload (Instructing the Model)
# We give it a clear task and feed it the context.
payload = {
    "contents": [{
        "parts": [{"text": f"Extract the following from this AAR:\n1. Units involved\n2. Primary obstacles\n3. Mission Outcome\n\nAAR TEXT:\n{aar_text}"}]
    }]
}

headers = {'Content-Type': 'application/json'}

# 4. Execute the API Call
print("Transmitting AAR to LLM endpoint for processing...")
# We append the API key to the URL parameters
response = requests.post(f"{API_URL}?key={API_KEY}", headers=headers, data=json.dumps(payload))

# 5. Parse and Output the Actionable Intelligence
if response.status_code == 200:
    result = response.json()
    # Navigate the JSON structure to extract the raw text
    ai_summary = result['candidates'][0]['content']['parts'][0]['text']
    
    print("\n=== AI GENERATED INTELLIGENCE SUMMARY ===")
    print(ai_summary)
else:
    print(f"API Error: {response.status_code}")
    print(response.text)

```


### Interpreting the Results

When you execute this script, the LLM utilizes its **Self-Attention** network to link "heavy rain" to "denied air support" as an obstacle, and summarizes the chaotic text into three clean bullet points exactly as instructed.

By wrapping this API call in a Python script, an intelligence officer could loop this code over a folder containing 5,000 separate AARs, generating a unified, structured database of mission outcomes in seconds.
