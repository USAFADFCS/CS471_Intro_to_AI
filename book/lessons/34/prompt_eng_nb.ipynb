## EX: Engineered Prompts via API

In this exercise, we will use Python to test the difference between a weak prompt and a highly engineered prompt using the Task/Format/Voice/Context framework. We will also implement a meta-prompt to force step-by-step reasoning.

* **Baseline Testing** Send a generic, weak prompt to the API to observe the model's default, unfiltered behavior.

* **Framework Application** Construct and send a highly structured prompt utilizing the Task, Format, Voice, and Context framework to generate a specific, tailored lesson plan.

* **Cognitive Forcing (Meta-Prompting)** Submit a complex logistics math problem accompanied by a step-by-step meta-prompt to observe how forcing the model to show its intermediate reasoning prevents hallucinations.

```python
import requests
import json

# Define the API connection (Gemini 2.5 Flash via REST)
API_URL = "https://generativelanguage.googleapis.com/v1beta/models/gemini-3-flash-preview:generateContent"
API_KEY = "YOUR_GOOGLE_KEY_HERE"

def call_llm(prompt):
    payload = {"contents": [{"parts": [{"text": prompt}]}]}
    headers = {'Content-Type': 'application/json'}
    response = requests.post(f"{API_URL}?key={API_KEY}", headers=headers, data=json.dumps(payload))
    if response.status_code == 200:
        return response.json()['candidates'][0]['content']['parts'][0]['text']
    return f"Error: {response.status_code}"

# 1. The Weak Prompt
weak_prompt = "Write something about military drones."
print("=== WEAK PROMPT RESPONSE ===")
print(call_llm(weak_prompt))
print("\n" + "="*40 + "\n")

# 2. The Engineered Prompt (Task, Format, Voice, Context)
engineered_prompt = """
[TASK] Reimagine the concept of military drones as a classroom exercise.
[FORMAT] Provide a 3-point lesson plan outline. 
[VOICE] Write in a highly enthusiastic, encouraging tone suitable for a middle-school science teacher.
[CONTEXT] The students have no prior knowledge of aerodynamics.
"""
print("=== ENGINEERED PROMPT RESPONSE ===")
print(call_llm(engineered_prompt))
print("\n" + "="*40 + "\n")

# 3. Forcing Step-by-Step Logic (Meta-Prompting)
logic_prompt = """
A drone leaves base at 0800 traveling 100mph due North. 
At 0930, it turns East and accelerates to 150mph. 
What is its distance from the base at 1100?

Take a deep breath and work on this problem step-by-step. Break down your math before giving the final answer.
"""
print("=== STEP-BY-STEP LOGIC RESPONSE ===")
print(call_llm(logic_prompt))
```

=== STEP-BY-STEP LOGIC RESPONSE ===
To find the distance of the drone from the base at 1100, we need to calculate the displacement for each leg of the journey and then use the Pythagorean theorem to find the total distance from the starting point.

### Step 1: Calculate the distance traveled North
The drone travels North from 0800 to 0930.
*   **Time elapsed:** 1 hour and 30 minutes (1.5 hours).
*   **Speed:** 100 mph.
*   **Distance North:** $100 \text{ mph} \times 1.5 \text{ hours} = 150 \text{ miles}$.

At 0930, the drone's position relative to the base is **(0, 150)**.

### Step 2: Calculate the distance traveled East
The drone turns East and travels from 0930 to 1100.
*   **Time elapsed:** 1 hour and 30 minutes (1.5 hours).
*   **Speed:** 150 mph.
*   **Distance East:** $150 \text{ mph} \times 1.5 \text{ hours} = 225 \text{ miles}$.

At 1100, the drone's position relative to the base is **(225, 150)**.

### Step 3: Calculate the straight-line distance from the base
The distance from the base $(0, 0)$ to the final position $(225, 150)$ forms a right-angled triangle where the legs are 225 miles and 150 miles. We use the Pythagorean theorem ($a^2 + b^2 = c^2$):

$$c = \sqrt{225^2 + 150^2}$$

To make the math easier, notice that both 225 and 150 are multiples of 75:
*   $225 = 75 \times 3$
*   $150 = 75 \times 2$

Now we can factor out the 75:
$$c = \sqrt{(75 \times 3)^2 + (75 \times 2)^2}$$
$$c = \sqrt{75^2(3^2 + 2^2)}$$
$$c = 75 \times \sqrt{9 + 4}$$
$$c = 75\sqrt{13}$$

### Final Calculation:
To get a numerical value:
*   $\sqrt{13} \approx 3.60555$
*   $75 \times 3.60555 \approx 270.416$

**Final Answer:**
The distance from the base is **$75\sqrt{13}$ miles**, which is approximately **270.42 miles**.
