# EX: Building a Basic AI Pipeline

In this exercise, we will simulate building a simple detection pipeline using Python and `scikit-learn`. We will link a data scaler (preprocessing) and a classifier (learning/inference) into a single unified object. Later we will look closer at what a Support Vector Machine is, for now just assume it is a mathmatical model that is a classifier, e.g., help us determine the class of a given input.

**Steps Performed:**

* **Generate synthetic sensor data** by representing friend/foe radar cross-sections.
* **Define a preprocessing step** to normalize the sensor data.
* **Select a mathmatical model** by applying a Support Vector Machine (SVM) classifier to act as our inference engine.
* **Create the pipeline** combine them into a *Pipeline* and execute the transformation from raw input to tactical output.

````python
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

# Step 1: Simulated raw radar cross-section data (Inputs)
# Features: [Speed (knots), Altitude (feet)]
X_raw_sensor = np.array([[400, 15000], [450, 18000], [800, 30000], [900, 35000]])
# Labels: 0 = Commercial (Friend), 1 = Fighter (Foe)
# Note: The first two sets of Speed and Altitude readings are from commercial aircraft, while the last two are from hostile fighter jets.
y_true_labels = np.array([0, 0, 1, 1])

# Step 2 & 3: Define the Pipeline components
tactical_pipeline = Pipeline([
    ('preprocessor', StandardScaler()), # Normalizes the data
    ('classifier', SVC(kernel='linear')) # Makes the classification
])

# Step 4: Train the pipeline and test a new input
tactical_pipeline.fit(X_raw_sensor, y_true_labels)

# New unknown sensor reading comes in: 850 knots at 32000 feet
unknown_contact = np.array([[850, 32000]])
prediction = tactical_pipeline.predict(unknown_contact)

if prediction[0] == 1:
    print("ALERT: Hostile Fighter Detected.")
else:
    print("LOG: Commercial traffic.")
````

In this simulation, our unknown_contact has a high speed (850 knots) and high altitude (32,000 feet).

**Preprocessing:** The StandardScaler first normalizes these raw values to match the mathematical scale of our training data, preventing larger numbers (like altitude) from dominating the calculation.

**Inference:** The SVC model (our inference engine) compares these scaled features against the mathematical boundary it learned during training. Because the features closely cluster with the known hostile profiles, the model categorizes it as a 1.

**Actuation/Output:** The final if/else block serves as our system's actuator, translating the raw mathematical output (1) into a human-readable tactical alert ("ALERT: Hostile Fighter Detected") for the commander's screen.
