# EX: API Prompt Evaluation

In this exercise, we will simulate passing data to an LLM via an API. We will observe how a weak, vague prompt leads to an inconsistent output (a potential hallucination), and how redesigning the input with strict constraints forces the AI system to output a reliable format.

**Steps Performed:**

* **Define a simulation function** that acts as our LLM's inference engine, evaluating how strict the input prompt is.

* **Provide the input data** as a raw intelligence intercept to the system.

* **Test the system** with a vague, Zero-Shot prompt and observe the rambling, hallucinated output (Test 1).

* **Redesign the input** using strict Prompt Engineering constraints (formatting, persona, and strict instructions) and observe the corrected data output (Test 2).


````python
# Note: In a live environment, you would use 'google-generativeai' or 'openai' packages.
#  We will introduce API calls in lessons 33 and 34, but for now we simulate the LLM's inference response based on prompt design.

def simulate_llm_inference(prompt):
    """Simulates an LLM evaluating a prompt and predicting the output."""
    prompt_lower = prompt.lower()
    
    # Check if the prompt has strict constraints (Input Design)
    if "format exactly as" in prompt_lower and "confidence:" in prompt_lower:
        return "[TARGET: Mobile Radar] [GRID: 45A] [CONFIDENCE: High]"
    
    # If the prompt is zero-shot and vague, the LLM hallucinates/rambles
    elif "intel" in prompt_lower:
        return "The intel suggests a radar. It might be mobile. I think it is in sector 45A. There could also be a tank battalion nearby, as radars usually accompany tanks."
    
    return "I am an AI assistant. How can I help?"

# --- Scenario ---
raw_intercept = "Convoy moving to sector 45A. Setting up scanning dish."

print("--- TEST 1: Vague Zero-Shot Input ---")
vague_prompt = f"Analyze this intel: {raw_intercept}"
output_1 = simulate_llm_inference(vague_prompt)
print(f"PROMPT: {vague_prompt}")
print(f"OUTPUT: {output_1}\n")

print("--- TEST 2: High-Constraint Input Design ---")
strict_prompt = f"""
You are a tactical intelligence AI. Extract the target type and grid location from the intercept.
Do not invent information. 
Format exactly as: [TARGET: ] [GRID: ] [CONFIDENCE: ]

Intercept: {raw_intercept}
"""
output_2 = simulate_llm_inference(strict_prompt)
print(f"PROMPT: {strict_prompt}")
print(f"OUTPUT: {output_2}")
````

In Test 1, the input was completely unconstrained. The LLM processed the text but acted like a generalized chatbot, rambling and eventually hallucinating the presence of a "tank battalion" just because radars and tanks are statistically linked in its training data.

In Test 2, we changed the input design. By defining a persona, explicitly forbidding hallucinations, and providing a strict output template, we mathematically forced the LLM's inference engine to abandon its generalized conversational mode and output a clean, parsable data string suitable for a military dashboard.
